### Q19. Create a list of 10 numbers and remove duplicates.

### Q34. Create a list and find common elements between two lists.

```python
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7]
common = []
for n in list1:
    if n in list2:
        common.append(n)
print("Common elements:", common)
```

### Q35. Create a list and remove all even numbers.

```python
nums = [1, 2, 3, 4, 5, 6, 7, 8]
result = []
for n in nums:
    if n % 2 != 0:
        result.append(n)
print("After removing even numbers:", result)
```

### Q36. Create a list and remove all odd numbers.

```python
nums = [1, 2, 3, 4, 5, 6, 7, 8]
result = []
for n in nums:
    if n % 2 == 0:
        result.append(n)
print("After removing odd numbers:", result)
```

### Q37. Create a list and check palindrome list.

### Q40. Create a list and demonstrate all list methods.

```python
my_list = [5, 3, 8]
my_list.append(1)
print("append:", my_list)
my_list.insert(1, 9)
print("insert:", my_list)
my_list.extend([7, 2])
print("extend:", my_list)
my_list.remove(8)
print("remove:", my_list)
print("pop:", my_list.pop())
print("count of 3:", my_list.count(3))
print("index of 9:", my_list.index(9))
my_list.sort()
print("sort:", my_list)
my_list.reverse()
print("reverse:", my_list)
new_list = my_list.copy()
print("copy:", new_list)
my_list.clear()
print("clear:", my_list)
```
