### Q4. Count the number of vowels in a string.

```python
text = input("Enter a string: ").lower()
count = 0
for ch in text:
    if ch in "aeiou":
        count += 1
print("Vowels:", count)
```

### Q5. Count the number of consonants in a string.

```python
text = input("Enter a string: ").lower()
count = 0
for ch in text:
    if ch.isalpha() and ch not in "aeiou":
        count += 1
print("Consonants:", count)
```

### Q31. Remove punctuation marks from a string.

```python
import string
text = input("Enter a string: ")
result = ""
for ch in text:
    if ch not in string.punctuation:
        result += ch
print("Without punctuation:", result)
```

### Q37. Find the longest word in a sentence.

```python
sentence = input("Enter a sentence: ")
longest = ""
for word in sentence.split():
    if len(word) > len(longest):
        longest = word
print("Longest word:", longest)
```

### Q38. Find the shortest word in a sentence.

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
shortest = words[0]
for word in words:
    if len(word) < len(shortest):
        shortest = word
print("Shortest word:", shortest)
```

### Q40. Count uppercase and lowercase letters separately.

```python
text = input("Enter a string: ")
upper = 0
lower = 0
for ch in text:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1
print("Uppercase letters:", upper)
print("Lowercase letters:", lower)
```
