## Mini Project 1: Smart Electricity Bill Calculator

Charges: first 100 units Rs3, next 100 units Rs5, above 200 units Rs7. Fixed service charge Rs150. Bill above Rs5000 = High Usage Bill.

```python
# Mini Project 1 - Electricity Bill

name = input("Enter customer name: ")
units = int(input("Enter units consumed: "))

if units < 0:
    print("Units cannot be negative!")
else:
    # calculating energy charge slab wise
    if units <= 100:
        energy_charge = units * 3
    elif units <= 200:
        # first 100 units at 3 and remaining at 5
        energy_charge = (100 * 3) + (units - 100) * 5
    else:
        # first 100 at 3, next 100 at 5, rest at 7
        energy_charge = (100 * 3) + (100 * 5) + (units - 200) * 7

    service_charge = 150
    final_bill = energy_charge + service_charge

    print("\n----- Electricity Bill -----")
    print("Customer Name   :", name)
    print("Units Consumed  :", units)
    print("Energy Charge   : Rs", energy_charge)
    print("Service Charge  : Rs", service_charge)
    print("Final Bill      : Rs", final_bill)

    if final_bill > 5000:
        print("Usage Message   : High Usage Bill")
    else:
        print("Usage Message   : Normal Usage")
```

**Test case:** inputs given = Ravi, 350

## Mini Project 2: Online Shopping Checkout

Discount by purchase amount (20% / 10% / 5% / none), premium customers get extra 5%. Delivery free if final amount is Rs2000 or more, otherwise Rs100.

```python
# Mini Project 2 - Online Shopping Checkout

name = input("Enter customer name: ")
amount = float(input("Enter purchase amount: "))
cust_type = input("Customer type (Regular/Premium): ").strip().capitalize()

if amount < 0:
    print("Purchase amount cannot be negative!")
elif cust_type != "Regular" and cust_type != "Premium":
    print("Invalid customer type! Enter Regular or Premium.")
else:
    # discount percent based on amount
    if amount >= 5000:
        discount_percent = 20
    elif amount >= 3000 and amount < 5000:
        discount_percent = 10
    elif amount >= 1000 and amount < 3000:
        discount_percent = 5
    else:
        discount_percent = 0

    # extra 5% for premium
    if cust_type == "Premium":
        discount_percent = discount_percent + 5

    discount_amount = amount * discount_percent / 100
    amount_after_discount = amount - discount_amount

    # delivery charge
    if amount_after_discount >= 2000:
        delivery = 0
    else:
        delivery = 100

    final_amount = amount_after_discount + delivery

    print("\n----- Checkout Summary -----")
    print("Customer Name    :", name)
    print("Discount Amount  : Rs", round(discount_amount, 2))
    print("Delivery Charge  : Rs", delivery)
    print("Final Payable    : Rs", round(final_amount, 2))
```

**Test case:** inputs given = Anita, 3500, premium

## Mini Project 3: ATM Withdrawal System

Withdrawal must be > 0, not more than balance, max Rs20000 per transaction. Show Low Balance Warning if balance falls below Rs1000.

```python
# Mini Project 3 - ATM Withdrawal

name = input("Enter account holder name: ")
balance = float(input("Enter current balance: "))
withdraw = float(input("Enter withdrawal amount: "))

print("\n----- ATM Transaction -----")
print("Account Holder :", name)

if withdraw <= 0:
    print("Transaction Status : Rejected")
    print("Reason : Withdrawal amount must be greater than 0")
else:
    if withdraw > balance:
        print("Transaction Status : Rejected")
        print("Reason : Insufficient balance")
    else:
        if withdraw > 20000:
            print("Transaction Status : Rejected")
            print("Reason : Maximum limit per transaction is Rs 20000")
        else:
            balance = balance - withdraw
            print("Transaction Status : Successful")
            print("Withdrawn Amount   : Rs", withdraw)
            print("Remaining Balance  : Rs", balance)
            if balance < 1000:
                print("Low Balance Warning")
```

**Test case:** inputs given = Suresh, 5000, 4200

## Mini Project 4: Student Scholarship Eligibility

Need percentage >= 75, income <= 4,00,000 and attendance >= 75. 90%+ = Full Scholarship, otherwise Partial. Else Not Eligible with reason.

```python
# Mini Project 4 - Scholarship Eligibility

name = input("Enter student name: ")
percentage = float(input("Enter percentage: "))
income = float(input("Enter annual family income: "))
attendance = float(input("Enter attendance percentage: "))

print("\n----- Scholarship Result -----")
print("Student Name :", name)

if percentage >= 75 and income <= 400000 and attendance >= 75:
    print("Eligibility Status : Eligible")
    if percentage >= 90:
        print("Scholarship Category : Full Scholarship")
    else:
        print("Scholarship Category : Partial Scholarship")
else:
    print("Eligibility Status : Not Eligible")
    print("Scholarship Category : None")
    # finding the reason(s)
    if percentage < 75:
        print("Reason : Percentage is below 75%")
    if income > 400000:
        print("Reason : Family income is more than 4,00,000")
    if attendance < 75:
        print("Reason : Attendance is below 75%")
```

**Test case:** inputs given = Neha, 92, 300000, 85

## Mini Project 5: Employee Performance Bonus

Bonus depends on experience and rating (1-5). Rating 5 = Excellent Performer. Invalid rating is rejected.

```python
# Mini Project 5 - Employee Bonus

name = input("Enter employee name: ")
salary = float(input("Enter monthly salary: "))
experience = int(input("Enter years of experience: "))
rating = int(input("Enter performance rating (1 to 5): "))

if rating < 1 or rating > 5:
    print("Invalid rating! Rating should be between 1 and 5.")
else:
    if experience >= 10 and rating >= 4:
        bonus_percent = 20
    elif experience >= 5 and rating >= 4:
        bonus_percent = 15
    elif experience >= 2 and rating >= 3:
        bonus_percent = 10
    else:
        # rating below 3 or experience too low
        bonus_percent = 0

    bonus_amount = salary * bonus_percent / 100
    final_salary = salary + bonus_amount

    print("\n----- Bonus Report -----")
    print("Employee Name    :", name)
    print("Bonus Percentage :", bonus_percent, "%")
    print("Bonus Amount     : Rs", bonus_amount)
    print("Final Salary     : Rs", final_salary)

    if rating == 5:
        print("Message          : Excellent Performer")
    elif rating >= 3:
        print("Message          : Good work")
    else:
        print("Message          : Needs improvement")
```

**Test case:** inputs given = Amit, 40000, 6, 5

## Mini Project 6: Hotel Room Booking System

Standard 2000, Deluxe 3500, Suite 6000 per night. 5+ nights = 10% off, 10+ nights = 20% off.

```python
# Mini Project 6 - Hotel Booking

guest = input("Enter guest name: ")
room_type = input("Enter room type (Standard/Deluxe/Suite): ").strip().capitalize()
nights = int(input("Enter number of nights: "))

if nights <= 0:
    print("Number of nights must be greater than 0!")
else:
    if room_type == "Standard":
        price = 2000
    elif room_type == "Deluxe":
        price = 3500
    elif room_type == "Suite":
        price = 6000
    else:
        price = 0

    if price == 0:
        print("Error: Invalid room type!")
    else:
        room_charge = price * nights

        if nights >= 10:
            discount_percent = 20
        elif nights >= 5:
            discount_percent = 10
        else:
            discount_percent = 0

        discount = room_charge * discount_percent / 100
        final_amount = room_charge - discount

        print("\n----- Booking Details -----")
        print("Guest Name      :", guest)
        print("Room Type       :", room_type)
        print("Nights          :", nights)
        print("Room Charge     : Rs", room_charge)
        print("Discount        : Rs", discount)
        print("Final Amount    : Rs", final_amount)
```

**Test case:** inputs given = Rahul, deluxe, 6

## Mini Project 7: Mobile Data Plan Bill

Basic 299 (10GB), Standard 499 (25GB), Premium 799 (50GB). Extra data Rs20/GB, surcharge Rs100 if extra data is more than 10GB.

```python
# Mini Project 7 - Mobile Data Bill

name = input("Enter customer name: ")
plan = input("Enter plan type (Basic/Standard/Premium): ").strip().capitalize()
extra_gb = float(input("Enter extra data used (GB): "))

if extra_gb < 0:
    print("Data usage cannot be negative!")
else:
    if plan == "Basic":
        base_charge = 299
    elif plan == "Standard":
        base_charge = 499
    elif plan == "Premium":
        base_charge = 799
    else:
        base_charge = 0

    if base_charge == 0:
        print("Invalid plan type!")
    else:
        extra_charge = extra_gb * 20

        if extra_gb > 10:
            surcharge = 100
        else:
            surcharge = 0

        final_bill = base_charge + extra_charge + surcharge

        print("\n----- Mobile Bill -----")
        print("Customer Name      :", name)
        print("Plan Type          :", plan)
        print("Base Plan Charge   : Rs", base_charge)
        print("Extra Data Charge  : Rs", extra_charge)
        print("Surcharge          : Rs", surcharge)
        print("Final Bill         : Rs", final_bill)
```

**Test case:** inputs given = Pooja, Standard, 12

## Mini Project 8: Movie Ticket Booking

Base price 200. Child below 12 gets 50% off, senior 60+ gets 30% off. Morning show Rs30 off, weekend Rs50 extra.

```python
# Mini Project 8 - Movie Ticket

name = input("Enter customer name: ")
age = int(input("Enter age: "))
show = input("Enter show type (Morning/Evening): ").strip().capitalize()
day = input("Enter day type (Weekday/Weekend): ").strip().capitalize()

if age < 0:
    print("Age cannot be negative!")
elif (show != "Morning" and show != "Evening") or (day != "Weekday" and day != "Weekend"):
    print("Invalid show type or day type!")
else:
    base_price = 200
    discount = 0

    # age based discount
    if age < 12:
        discount = base_price * 50 / 100
    elif age >= 60:
        discount = base_price * 30 / 100

    # morning show extra discount
    if show == "Morning":
        discount = discount + 30

    # weekend surcharge
    if day == "Weekend":
        surcharge = 50
    else:
        surcharge = 0

    final_price = base_price - discount + surcharge

    print("\n----- Ticket Details -----")
    print("Customer Name         :", name)
    print("Original Ticket Price : Rs", base_price)
    print("Discount              : Rs", discount)
    print("Weekend Surcharge     : Rs", surcharge)
    print("Final Ticket Price    : Rs", final_price)
```

**Test case:** inputs given = Karan, 65, morning, weekend

## Mini Project 9: Insurance Premium Calculator

Base premium 10000. Age above 50 +5000, high risk job +7000, medium risk +3000, smoker +5000. Under 18 not eligible.

```python
# Mini Project 9 - Insurance Premium

name = input("Enter customer name: ")
age = int(input("Enter age: "))
risk = input("Enter occupation risk (Low/Medium/High): ").strip().capitalize()
smoker = input("Smoker? (Yes/No): ").strip().capitalize()

print("\n----- Insurance Details -----")
print("Customer Name :", name)

if age < 18:
    print("Eligibility   : Not Eligible (age is below 18)")
else:
    if risk != "Low" and risk != "Medium" and risk != "High":
        print("Invalid occupation risk!")
    elif smoker != "Yes" and smoker != "No":
        print("Smoker should be Yes or No!")
    else:
        base_premium = 10000
        additional = 0

        if age > 50:
            additional = additional + 5000

        if risk == "High":
            additional = additional + 7000
        elif risk == "Medium":
            additional = additional + 3000

        if smoker == "Yes":
            additional = additional + 5000

        final_premium = base_premium + additional

        print("Eligibility        : Eligible")
        print("Base Premium       : Rs", base_premium)
        print("Additional Charges : Rs", additional)
        print("Final Premium      : Rs", final_premium)

        if additional >= 10000:
            print("Risk Message       : High Risk Policy")
        else:
            print("Risk Message       : Normal Policy")
```

**Test case:** inputs given = Vikas, 55, Medium, yes

## Mini Project 10: Restaurant Bill and Service Charge

Discount by bill amount (0/5/10/15%), members get extra 5%. Then add 5% service charge on the discounted amount.

```python
# Mini Project 10 - Restaurant Bill

name = input("Enter customer name: ")
bill = float(input("Enter food bill amount: "))
member = input("Membership (Yes/No): ").strip().capitalize()

if bill < 0:
    print("Bill amount cannot be negative!")
elif member != "Yes" and member != "No":
    print("Membership should be Yes or No!")
else:
    # discount percent as per bill
    if bill >= 5000:
        discount_percent = 15
    elif bill >= 3000:
        discount_percent = 10
    elif bill >= 1000:
        discount_percent = 5
    else:
        discount_percent = 0

    # extra 5% for members
    if member == "Yes":
        discount_percent = discount_percent + 5

    discount = bill * discount_percent / 100
    after_discount = bill - discount

    # service charge is added after discount
    service_charge = after_discount * 5 / 100
    final_amount = after_discount + service_charge

    print("\n----- Restaurant Bill -----")
    print("Customer Name    :", name)
    print("Original Bill    : Rs", bill)
    print("Discount         : Rs", round(discount, 2))
    print("Service Charge   : Rs", round(service_charge, 2))
    print("Final Payable    : Rs", round(final_amount, 2))

    if final_amount >= 5000:
        print("Message          : Thank you for being a valued customer")
```

**Test case:** inputs given = Sneha, 5200, yes
