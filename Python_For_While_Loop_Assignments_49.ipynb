### Q12. Reverse a number using while loop.

```python
num = int(input("Enter a number: "))
rev = 0
while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10
print("Reversed number:", rev)
```

### Q13. Check whether a number is palindrome.

```python
num = int(input("Enter a number: "))
original = num
rev = 0
while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10
if original == rev:
    print(original, "is a palindrome")
else:
    print(original, "is not a palindrome")
```

### Q19. Check whether a number is prime.

```python
num = int(input("Enter a number: "))
is_prime = True
if num < 2:
    is_prime = False
else:
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
if is_prime:
    print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```

### Q28. Check whether a number is Armstrong.

```python
num = int(input("Enter a number: "))
temp = num
total = 0
digits = len(str(num))
while temp > 0:
    digit = temp % 10
    total = total + digit ** digits
    temp = temp // 10
if total == num:
    print(num, "is an Armstrong number")
else:
    print(num, "is not an Armstrong number")
```

### Q48. Check whether number is strong number.

```python
num = int(input("Enter a number: "))
temp = num
total = 0
while temp > 0:
    digit = temp % 10
    fact = 1
    for i in range(1, digit + 1):
        fact = fact * i
    total = total + fact
    temp = temp // 10
if total == num:
    print(num, "is a strong number")
else:
    print(num, "is not a strong number")
```

### Q49. Display menu-driven program using while loop.

```python
choice = 0
while choice != 4:
    print("\n1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Exit")
    choice = int(input("Enter your choice: "))
    if choice == 1 or choice == 2 or choice == 3:
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))
        if choice == 1:
            print("Result:", a + b)
        elif choice == 2:
            print("Result:", a - b)
        else:
            print("Result:", a * b)
    elif choice == 4:
        print("Exiting program")
    else:
        print("Invalid choice")
```
