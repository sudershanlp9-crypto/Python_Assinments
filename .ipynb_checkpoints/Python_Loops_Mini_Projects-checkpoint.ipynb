## Mini Project 1: Daily Expense Tracker

Enter expense for each day, find total, average, highest day. If total is more than Rs5000 show Budget Alert.

```python
# Mini Project 1 - Daily Expense Tracker

days = int(input("Enter number of days: "))

if days <= 0:
    print("Number of days must be greater than 0")
else:
    total = 0
    highest = 0

    for i in range(1, days + 1):
        expense = float(input("Enter expense for day " + str(i) + ": "))
        # if user enters negative number ask again
        while expense < 0:
            print("Expense cannot be negative, try again")
            expense = float(input("Enter expense for day " + str(i) + ": "))

        total = total + expense
        if expense > highest:
            highest = expense

    average = total / days

    print("\n----- Expense Summary -----")
    print("Total Expense          : Rs", total)
    print("Average Daily Expense  : Rs", round(average, 2))
    print("Highest Daily Expense  : Rs", highest)

    if total > 5000:
        print("Budget Status          : Budget Alert")
    else:
        print("Budget Status          : Within Budget")
```

**Test case:** inputs given = 4, 1200, 800, 2500, 900

## Mini Project 2: Student Marks and Grade Report

Grades: 75+ A, 60-74 B, 50-59 C, 40-49 D, below 40 Fail. Marks outside 0-100 are rejected.

```python
# Mini Project 2 - Student Marks and Grade Report

n = int(input("Enter number of students: "))

total_marks = 0
pass_count = 0
fail_count = 0

for i in range(n):
    name = input("Enter name of student " + str(i + 1) + ": ")
    marks = float(input("Enter marks out of 100: "))

    # reject marks outside 0-100
    while marks < 0 or marks > 100:
        print("Invalid marks! Enter between 0 and 100")
        marks = float(input("Enter marks out of 100: "))

    if marks >= 75:
        grade = "A"
    elif marks >= 60:
        grade = "B"
    elif marks >= 50:
        grade = "C"
    elif marks >= 40:
        grade = "D"
    else:
        grade = "Fail"

    if grade == "Fail":
        fail_count = fail_count + 1
    else:
        pass_count = pass_count + 1

    total_marks = total_marks + marks
    print(name, "- Grade:", grade)

if n > 0:
    class_average = total_marks / n
    print("\n----- Class Summary -----")
    print("Class Average :", round(class_average, 2))
    print("Pass Count    :", pass_count)
    print("Fail Count    :", fail_count)
```

**Test case:** inputs given = 4, Amit, 82, Riya, 55, Sam, 105, 35, Tina, 68

## Mini Project 3: ATM PIN Verification

Sample PIN is stored in the program. User gets 3 attempts. After 3 wrong attempts the card is locked.

```python
# Mini Project 3 - ATM PIN Verification

correct_pin = 1234
attempts = 0
access = False

while attempts < 3:
    pin = int(input("Enter your PIN: "))
    attempts = attempts + 1

    if pin == correct_pin:
        access = True
        break
    else:
        print("Wrong PIN! Attempts left:", 3 - attempts)

print("\n----- Result -----")
if access:
    print("Access Status : Access Granted")
else:
    print("Access Status : Access Denied")
    print("Card Temporarily Locked")

print("Attempts Used :", attempts)
```

**Test case:** inputs given = 1111, 2222, 1234

## Mini Project 4: Shopping Cart Total

Enter price and quantity for each product. 10% discount if subtotal is Rs3000 or more. Delivery Rs100 if discounted amount is below Rs1500.

```python
# Mini Project 4 - Shopping Cart

n = int(input("Enter number of products: "))
subtotal = 0

for i in range(1, n + 1):
    print("\nProduct", i)
    price = float(input("Enter price: "))
    quantity = int(input("Enter quantity: "))

    # negative values are not allowed
    while price < 0 or quantity < 0:
        print("Price and quantity cannot be negative, enter again")
        price = float(input("Enter price: "))
        quantity = int(input("Enter quantity: "))

    item_total = price * quantity
    print("Item subtotal: Rs", item_total)
    subtotal = subtotal + item_total

# discount
if subtotal >= 3000:
    discount = subtotal * 10 / 100
else:
    discount = 0

after_discount = subtotal - discount

# delivery
if after_discount < 1500:
    delivery = 100
else:
    delivery = 0

final_amount = after_discount + delivery

print("\n----- Bill -----")
print("Subtotal        : Rs", subtotal)
print("Discount        : Rs", discount)
print("Delivery Charge : Rs", delivery)
print("Final Payable   : Rs", final_amount)
```

**Test case:** inputs given = 3, 1000, 2, -50, 500, 50, 10, 800, 1

## Mini Project 5: Number Guessing Game

Secret number is 37. User gets 5 guesses and hints Too low / Too high.

```python
# Mini Project 5 - Number Guessing Game

secret = 37
attempts = 0
won = False

while attempts < 5:
    guess = int(input("Guess the number: "))
    attempts = attempts + 1

    if guess == secret:
        won = True
        break
    elif guess < secret:
        print("Too low")
    else:
        print("Too high")

print("\n----- Game Over -----")
if won:
    print("You win! You guessed the number.")
else:
    print("You lose! The secret number was", secret)

print("Attempts Used:", attempts)
```

**Test case:** inputs given = 10, 50, 40, 37

## Mini Project 6: Electricity Usage Analyzer

Enter units for each day. Find total, average, days above 10 units, maximum. Total above 300 = High Usage.

```python
# Mini Project 6 - Electricity Usage Analyzer

days = int(input("Enter number of days: "))

total = 0
maximum = 0
high_days = 0

for i in range(1, days + 1):
    units = float(input("Enter units used on day " + str(i) + ": "))

    total = total + units

    if units > 10:
        high_days = high_days + 1

    if units > maximum:
        maximum = units

if days > 0:
    average = total / days

    print("\n----- Usage Report -----")
    print("Total Units          :", total)
    print("Average Usage        :", round(average, 2))
    print("Days with > 10 units :", high_days)
    print("Maximum Daily Usage  :", maximum)

    if total > 300:
        print("Usage Status         : High Usage")
    else:
        print("Usage Status         : Normal Usage")
else:
    print("Days should be more than 0")
```

**Test case:** inputs given = 5, 8, 12, 15, 9, 11

## Mini Project 7: Employee Payroll Summary

Overtime Rs200/hour. Gross = basic + overtime. Deduct 10% if gross >= 50000 otherwise 5%. Find net salary and total payroll.

```python
# Mini Project 7 - Payroll Summary

n = int(input("Enter number of employees: "))
total_payroll = 0

for i in range(1, n + 1):
    print("\nEmployee", i)
    name = input("Enter employee name: ")
    basic = float(input("Enter basic salary: "))
    overtime_hours = float(input("Enter overtime hours: "))

    overtime_pay = overtime_hours * 200
    gross = basic + overtime_pay

    if gross >= 50000:
        deduction = gross * 10 / 100
    else:
        deduction = gross * 5 / 100

    net = gross - deduction
    total_payroll = total_payroll + net

    print("Name   :", name)
    print("Gross  : Rs", gross)
    print("Deduction : Rs", deduction)
    print("Net Salary: Rs", net)

print("\n----- Payroll Summary -----")
print("Employee Count     :", n)
print("Total Net Payroll  : Rs", total_payroll)
```

**Test case:** inputs given = 3, Amit, 45000, 10, Riya, 30000, 5, Sam, 60000, 0

## Mini Project 8: Quiz Application

5 questions stored in lists. 1 point per correct answer. Excellent at 80%+, Good at 60-79%, Keep Practicing below 60%.

```python
# Mini Project 8 - Quiz Application

# questions and answers stored in lists
questions = [
    "What is the capital of India?",
    "How many days are there in a week?",
    "Which language are we learning here? (Python/Java)",
    "What is 5 + 7?",
    "Which planet is called the Red Planet?"
]
answers = ["delhi", "7", "python", "12", "mars"]

score = 0

for i in range(len(questions)):
    print("Q" + str(i + 1) + ".", questions[i])
    user_answer = input("Your answer: ").strip().lower()

    if user_answer == answers[i]:
        print("Correct!")
        score = score + 1
    else:
        print("Wrong! Correct answer is", answers[i])

percentage = (score / len(questions)) * 100

print("\n----- Quiz Result -----")
print("Score      :", score, "/", len(questions))
print("Percentage :", percentage, "%")

if percentage >= 80:
    print("Performance: Excellent")
elif percentage >= 60:
    print("Performance: Good")
else:
    print("Performance: Keep Practicing")
```

**Test case:** inputs given = Delhi, 7, Python, 13, Mars

## Mini Project 9: Bank Transaction History

Process deposits and withdrawals. Withdraw only if balance is enough. Count deposits, successful and rejected withdrawals. Warn if final balance is below Rs1000.

```python
# Mini Project 9 - Bank Transaction History

balance = float(input("Enter opening balance: "))
n = int(input("Enter number of transactions: "))

deposit_count = 0
withdraw_success = 0
withdraw_rejected = 0

for i in range(1, n + 1):
    print("\nTransaction", i)
    t_type = input("Enter type (deposit/withdraw): ").strip().lower()
    amount = float(input("Enter amount: "))

    if amount <= 0:
        print("Status: Invalid amount")
    elif t_type == "deposit":
        balance = balance + amount
        deposit_count = deposit_count + 1
        print("Status: Deposit successful")
    elif t_type == "withdraw":
        if amount <= balance:
            balance = balance - amount
            withdraw_success = withdraw_success + 1
            print("Status: Withdrawal successful")
        else:
            withdraw_rejected = withdraw_rejected + 1
            print("Status: Withdrawal rejected (insufficient funds)")
    else:
        print("Status: Invalid transaction type")

print("\n----- Account Summary -----")
print("Final Balance                :", balance)
print("Deposit Count                :", deposit_count)
print("Successful Withdrawal Count  :", withdraw_success)
print("Rejected Withdrawal Count    :", withdraw_rejected)

if balance < 1000:
    print("Warning: Balance is below Rs 1000")
```

**Test case:** inputs given = 2000, 4, deposit, 1500, withdraw, 5000, withdraw, 2500, withdraw, 300

## Mini Project 10: Multiplication Table and Number Analyzer

Print table of a number from 1 to 10, then count even/odd numbers and sum of a range.

```python
# Mini Project 10 - Multiplication Table and Number Analyzer

num = int(input("Enter a positive integer for the table: "))

if num <= 0:
    print("Please enter a positive integer")
else:
    print("\nMultiplication table of", num)
    for i in range(1, 11):
        print(num, "x", i, "=", num * i)

    start = int(input("\nEnter start of range: "))
    end = int(input("Enter end of range: "))

    if start > end:
        print("Start cannot be greater than end!")
    else:
        even_count = 0
        odd_count = 0
        total = 0

        for n in range(start, end + 1):
            if n % 2 == 0:
                even_count = even_count + 1
            else:
                odd_count = odd_count + 1
            total = total + n

        print("\n----- Range Analysis -----")
        print("Even Count   :", even_count)
        print("Odd Count    :", odd_count)
        print("Sum of Range :", total)
```

**Test case:** inputs given = 7, 1, 10
