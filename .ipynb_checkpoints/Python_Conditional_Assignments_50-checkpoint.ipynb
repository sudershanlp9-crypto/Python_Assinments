### Q23. Write a program to calculate electricity bill based on units consumed.

```python
units = int(input("Enter units consumed: "))
if units <= 100:
    bill = units * 3
elif units <= 200:
    bill = 100 * 3 + (units - 100) * 5
else:
    bill = 100 * 3 + 100 * 5 + (units - 200) * 7
print("Electricity bill: Rs", bill)
```

### Q40. Write a program to check whether a password is valid using nested if.

```python
password = input("Enter password: ")
if len(password) >= 8:
    if password.isalnum():
        if not password.isalpha() and not password.isdigit():
            print("Valid password")
        else:
            print("Invalid: use both letters and numbers")
    else:
        print("Invalid: no special characters allowed")
else:
    print("Invalid: password must be at least 8 characters")
```

### Q41. Write a program to print weekday name using match case.

```python
day = int(input("Enter day number (1-7): "))
match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid day number")
```

### Q42. Write a program to perform calculator operations using match case.

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
op = input("Enter operator (+, -, *, /): ")
match op:
    case "+":
        print("Result:", a + b)
    case "-":
        print("Result:", a - b)
    case "*":
        print("Result:", a * b)
    case "/":
        if b == 0:
            print("Cannot divide by zero")
        else:
            print("Result:", a / b)
    case _:
        print("Invalid operator")
```

### Q43. Write a program to print month days using match case.

```python
month = int(input("Enter month number (1-12): "))
match month:
    case 1 | 3 | 5 | 7 | 8 | 10 | 12:
        print("This month has 31 days")
    case 4 | 6 | 9 | 11:
        print("This month has 30 days")
    case 2:
        print("This month has 28 days (29 in leap year)")
    case _:
        print("Invalid month number")
```

### Q44. Write a program to check traffic signal actions using match case.

```python
signal = input("Enter signal colour (red/yellow/green): ").lower()
match signal:
    case "red":
        print("Stop")
    case "yellow":
        print("Get ready")
    case "green":
        print("Go")
    case _:
        print("Invalid signal")
```

### Q50. Write a program to build a simple ATM menu using a match case.

```python
balance = 10000
print("1. Check Balance")
print("2. Withdraw")
print("3. Deposit")
print("4. Exit")
choice = int(input("Enter your choice: "))
match choice:
    case 1:
        print("Your balance is Rs", balance)
    case 2:
        amount = int(input("Enter amount to withdraw: "))
        if amount <= balance:
            balance = balance - amount
            print("Withdrawal successful. New balance: Rs", balance)
        else:
            print("Insufficient balance")
    case 3:
        amount = int(input("Enter amount to deposit: "))
        balance = balance + amount
        print("Deposit successful. New balance: Rs", balance)
    case 4:
        print("Thank you for using the ATM")
    case _:
        print("Invalid choice")
```
